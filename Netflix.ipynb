```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("netflix_titles.csv..csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["rating"].unique()
```

```python
mask = df['rating'].isin(['74 min', '84 min', '66 min'])
print(mask.sum())        # tells you how many rows are True
print(df[mask])          # shows you those actual rows
```

```python
df.loc[mask, 'duration'] = df.loc[mask, 'rating']
df.loc[mask, 'duration'] = df.loc[mask, 'rating']
df.loc[mask, 'rating'] = None
```

```python
df.info()
```

```python
df["director"]=df["director"].fillna("Unknown")
df["country"]=df["country"].fillna("Unknown")
df["cast"]=df["cast"].fillna("Unknown")
```

```python
df.info()
```

```python
rm=df["rating"].mode()[0]
rm
```

```python
df["rating"]=df["rating"].fillna(rm)
```

```python
df.info()
```

```python
df = df.dropna(subset=['date_added', 'duration'])
df.info()
```

```python
df.head()
```

```python
df["date_added"]=pd.to_datetime(df["date_added"].str.strip())
```

```python
df.info()
```

```python
df["Year"]=df["date_added"].dt.year
df["Month_name"]=df["date_added"].dt.month_name()
```

```python
df.shape
```

```python
df[['date_added', 'Year', 'Month_name']].head()
```

```python
df_genres=df[["show_id", "listed_in"]].copy()
df_genres["listed_in"]=df_genres["listed_in"].str.split(",")
df_genres=df_genres.explode("listed_in")
df_genres.rename(columns={'listed_in': 'genre'}, inplace=True)
```

```python
df_genres.head()
```

```python
df_country=df[["show_id", "country"]].copy()
df_country["country"]=df_country["country"].str.split()
df_country=df_country.explode("country")
df_country.head()
```

```python
#exporting
df.to_csv("netflix_cleaned_data.csv", index=False)
df_genres.to_csv('netflix_genres.csv', index=False)
df_country.to_csv('netflix_country.csv', index=False)
```

```python
df['cast'] = df['cast'].astype(str).str.replace('\\', '', regex=False)
df['listed_in'] = df['listed_in'].astype(str).str.replace('\\', '', regex=False)
df['description'] = df['description'].astype(str).str.replace('\\', '', regex=False)
df['title'] = df['title'].astype(str).str.replace('\\', '', regex=False)

df.to_csv('netflix_cleaned_data2.csv', index=False)
```

```python
df_countries = df[['show_id', 'country']].copy()
df_countries['country'] = df_countries['country'].str.split(', ')
df_countries = df_countries.explode('country')
df_countries['country'] = df_countries['country'].str.strip()

df_countries.to_csv('netflix_countries2.csv', index=False)
```
